```python
import sympy as sp

x = sp.Function("x")
t, F, m= sp.symbols("t F m")


newton_2 = sp.Eq(F, m * sp.diff(x(t), t, t))
newton_2


```

```python
k = sp.symbols("k")

mass_on_spring = newton_2.subs(F, -k * x(t))
mass_on_spring
```

```python
a = sp.symbols("a")

trial_soln = mass_on_spring.subs(x(t), sp.exp(a * t))
trial_soln
```

```python
trial_soln.simplify()
```

```python
k_soln = sp.solve(trial_soln.simplify())[0][k]

sp.expand(k_soln)
```

```python
sp.dsolve(mass_on_spring)
```
